```
arr=np.zeros((2, 3))   # Nollar bilan to'ldirilgan (2x3) massiv
print("arr",arr)

arr2=np.ones((3, 3))   # Birlar bilan to'ldirilgan (3x3) massiv
print("arr2",arr2)

arr3=np.arange(0, 10, 2)     # [0, 2, 4, 6, 8] (range kabi)
print("arr3",arr3)

arr4=np.linspace(0, 10, 5)    # 0 dan 1 gacha oraliqni 5 ta teng qismga bo'lish
print("arr4",arr4)

arr5=np.random.rand(2, 5) # 0 va 1 oralig'idagi tasodifiy sonlar
print("arr5",arr5)
```

```
# flaten()
matritsa = np.array([[1, 2, 3],
                     [4, 5, 6]])
tekis = matritsa.flatten()
print(tekis)
print(tekis.shape)
```

```
A = np.array([[1, 2, 3],
              [4, 5, 6]])
print(A.shape)  # (2, 3)
# Transponirlash (.T yoki np.transpose(A))
A_T = A.T
print(A_T)
# Natija:
# [[1, 4],
#  [2, 5],
#  [3, 6]]
print(A_T.shape)  # (3, 2)
```

```
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

natija = np.dot(v1, v2)  # yoki: v1 @ v2
print(natija)  # 32
```
